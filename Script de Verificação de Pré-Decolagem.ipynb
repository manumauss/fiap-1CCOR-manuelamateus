```python
# Script de Verificação de Pré-Decolagem - Missão Aurora Siger

print("="*50)
print(" PAINEL DE CONTROLE - NAVE AURORA SIGER ")
print("="*50)
print("Insira os dados da telemetria atual para avaliação:\n")

try:
    # 1. Captura de Dados
    temp_interna = float(input("Temperatura Interna (°C): "))
    temp_externa = float(input("Temperatura Externa (°C): "))
    integridade = int(input("Integridade Estrutural (1 para Íntegra, 0 para Comprometida): "))
    energia = float(input("Nível de Energia (%): "))
    pressao_tanques = float(input("Pressão dos Tanques (kPa): "))
    status_modulos = input("Status dos Módulos Críticos (OK / FALHA): ").strip().upper()

    print("\n" + "="*50)
    print(" PROCESSANDO TELEMETRIA... ")
    print("="*50 + "\n")

    # 2. Lógica de Avaliação de Variáveis
    falha_critica = False
    alertas = 0
    mensagens = []

    # Verificação de Condições Absolutas
    if integridade == 0:
        falha_critica = True
        mensagens.append("[CRÍTICO] Falha estrutural detectada no casco.")
    if status_modulos != "OK":
        falha_critica = True
        mensagens.append("[CRÍTICO] Os módulos vitais não estão respondendo corretamente.")
    if not (-150.0 <= temp_externa <= 200.0): # Tolerância máxima absoluta
        falha_critica = True
        mensagens.append("[CRÍTICO] Temperatura externa em níveis catastróficos para a blindagem.")

    # Verificação de Condições com Margem de Tolerância
    if not (18.0 <= temp_interna <= 26.0):
        alertas += 1
        mensagens.append("[ALERTA] Temperatura interna fora da zona de conforto da tripulação.")
        
    if energia < 85:
        if energia >= 70:
            alertas += 1
            mensagens.append("[ALERTA] Energia abaixo do ideal. A autonomia pós-decolagem será reduzida.")
        else:
            falha_critica = True
            mensagens.append("[CRÍTICO] Energia insuficiente para o ciclo de ignição.")
            
    if not (350.0 <= pressao_tanques <= 450.0):
        if 320.0 <= pressao_tanques <= 480.0:
            alertas += 1
            mensagens.append("[ALERTA] Pressão dos tanques com leve instabilidade.")
        else:
            falha_critica = True
            mensagens.append("[CRÍTICO] Risco de ruptura ou falha de combustão devido à pressão anômala.")

    # 3. Exibição do Diagnóstico
    for msg in mensagens:
        print(msg)
    
    if not mensagens:
        print("Todos os sistemas operando dentro dos parâmetros nominais.")

    print("\n>>> DECISÃO DO SISTEMA <<<")

    # 4. Determinação do Status da Decolagem
    if falha_critica:
        print("STATUS: INVIÁVEL - DECOLAGEM ABORTADA")
        print("Ação: Acionar equipes de manutenção imediatamente.")
    elif alertas >= 2:
        print("STATUS: INSEGURA - DECOLAGEM SUSPENSA")
        print("Ação: Necessária revisão manual do comandante antes de prosseguir.")
    elif alertas == 1:
        print("STATUS: VIÁVEL - DECOLAGEM AUTORIZADA COM RESSALVAS")
        print("Ação: Monitoramento intensivo do parâmetro em alerta durante a subida.")
    else:
        print("STATUS: SEGURA - PRONTO PARA DECOLAR")
        print("Ação: Iniciar contagem regressiva.")

except ValueError:
    print("\n[ERRO] Tipo de dado inválido. Por favor, reinicie o sistema e digite apenas números onde for solicitado (ex: 22.5).")
```
